**2.2 Loading and Organizing Image Dataset**
*   Explanation of the directory structure and loading images:



```python
image_directory=os.path.join('/content/drive/MyDrive/', 'Brain_Tumor_DataSet/datasets/')
no_tumor_images=os.listdir(image_directory+ 'no/')
yes_tumor_images=os.listdir(image_directory+ 'yes/')

print('No Tumor: ', len(no_tumor_images))
print('Tumor: ',len(yes_tumor_images))
```

**2.3 Image Preprocessing**
*   Resizing and normalizing images:

```python
dataset=[]
label=[]

INPUT_SIZE=244

#for images labeled with no
for image_name in no_tumor_images:
    image=cv2.imread(image_directory+'no/'+image_name)
    image=Image.fromarray(image,'RGB')
    image=image.resize((INPUT_SIZE,INPUT_SIZE))
    dataset.append(np.array(image))
    label.append(0)

#for images labelled with yes
for image_name in yes_tumor_images:
    image=cv2.imread(image_directory+'yes/'+image_name)
    image=Image.fromarray(image,'RGB')
    image=image.resize((INPUT_SIZE,INPUT_SIZE))
    dataset.append(np.array(image))
    label.append(1)

dataset = np.array(dataset)
label = np.array(label)
```

**2.4 Splitting Dataset**



*   Splitting the dataset into training and testing:



```python
X_train, X_test, y_train, y_test = train_test_split(dataset, label, test_size=0.2, random_state=0)
X_train = X_train/255.0
X_test = X_test/255.0
```

**2.5 One-Hot Encoding of Labels**



*   Converting labels to categorical format:


```python
# Convert target values to one-hot vectors
y_train_encoded = to_categorical(y_train, num_classes=2)
y_test_encoded = to_categorical(y_test, num_classes=2)
```

```python
from graphviz import Digraph
from IPython.display import Image

def create_flowchart():
    # Initialize the graph object
    flowchart = Digraph(comment="Predict Multiple Images Flowchart", format='png')

    # Add nodes (steps in the process)
    flowchart.node('A', 'Start: Provide list of image paths (img_paths)')
    flowchart.node('B', 'Load & preprocess each image')
    flowchart.node('C', 'Resize to (INPUT_SIZE, INPUT_SIZE)')
    flowchart.node('D', 'Convert image to array & normalize (img / 255)')
    flowchart.node('E', 'Make prediction using model.predict(img)')
    flowchart.node('F', 'Store prediction probabilities')
    flowchart.node('G', 'Plot image and prediction')
    flowchart.node('H', 'Repeat for next image')
    flowchart.node('I', 'Show final plot of all predictions')

    # Add edges (flow connections)
    flowchart.edge('A', 'B')
    flowchart.edge('B', 'C')
    flowchart.edge('C', 'D')
    flowchart.edge('D', 'E')
    flowchart.edge('E', 'F')
    flowchart.edge('F', 'G')
    flowchart.edge('G', 'H')
    flowchart.edge('H', 'I')
    flowchart.edge('I', 'A')

    # Render and save the flowchart
    flowchart.render('flowchart-predict-multiple', format='png')

    # Display the image in the notebook
    return Image(filename='flowchart-predict-multiple.png')

# Create and display the flowchart
create_flowchart()
```

# 4. Model Building

**4.1 Model Architecture**

Explanation of the CNN architecture:
*   Input layer
*   Convolutional layers
*   Pooling layers
*   Dense layers
*   Dropout for regularization


Code:

```python
model=Sequential()

# Add an Input layer with the desired input shape
model.add(Input(shape=(INPUT_SIZE, INPUT_SIZE, 3)))

model.add(Conv2D(32, (3,3),activation='relu'))
model.add(MaxPooling2D(pool_size=(2,2)))

model.add(Conv2D(32, (3,3),activation='relu', kernel_initializer='he_uniform'))
model.add(MaxPooling2D(pool_size=(2,2)))

model.add(Conv2D(64, (3,3),activation='relu', kernel_initializer='he_uniform'))
model.add(MaxPooling2D(pool_size=(2,2)))

model.add(Flatten())
model.add(Dense(64,activation='relu'))
model.add(Dropout(0.5))
model.add(Dense(2,activation='softmax'))
```


**4.2 Model Compilation**



*   Setting loss function, optimizer, and metrics:


```python
model.compile(loss='categorical_crossentropy',optimizer='adam', metrics=['accuracy'])
```

**4.3 Model Summary**
*   Explanation of the model's layers and trainable parameters:


```python
model.summary()
```

4.4 Model Structure

```python
from diagrams import Diagram
from diagrams.aws.compute import EC2
from diagrams.aws.storage import S3
from IPython.display import Image

def create_model_structure():
    # Define the diagram and filename
    with Diagram("Model Structure", show=False, filename="/tmp/model_structure", direction="TB"):
        # Input Layer
        input_layer = EC2("Input Layer\n(224x224x3)")

        # Layers
        conv1 = EC2("Conv2D\n(32 filters)")
        pool1 = S3("MaxPooling\n(2x2)")
        conv2 = EC2("Conv2D\n(64 filters)")
        fc = EC2("Dense Layer\n(128 units)")
        output_layer = EC2("Output\n(Softmax)")

        # Connections between the layers
        input_layer >> conv1 >> pool1 >> conv2 >> fc >> output_layer

# Create the diagram
create_model_structure()

# Display the diagram image in Colab
Image("/tmp/model_structure.png")
```

# 5. Model Training

**5.1 Setting Callbacks**
*   Using EarlyStopping and ReduceLROnPlateau to improve training:


```python
# Stop training if loss doesn't keep decreasing.
model_es = EarlyStopping(monitor='loss', min_delta=1e-9, patience=10, verbose=True)
model_rlr = ReduceLROnPlateau(monitor='val_loss', factor=0.3, patience=8, verbose=True)

```

**5.2 Training the Model**
*   Training the model with data and callbacks:


```python
history = model.fit(X_train, y_train_encoded, batch_size=16, epochs=30, verbose=1, shuffle=False ,
                     validation_data=(X_test, y_test_encoded), callbacks=[model_es, model_rlr])
```

**5.3 Plotting Training History**


*   Visualizing accuracy and loss graphs:


```python
fig, ax = plt.subplots(1, 2, figsize=(10, 5))
# Plot training and validation accuracy
ax[0].plot(history.history['accuracy'])
ax[0].plot(history.history['val_accuracy'])
ax[0].set_title('Model Accuracy')
ax[0].set_xlabel('Epoch')
ax[0].set_ylabel('Accuracy')
ax[0].legend(['Train', 'Validation'], loc='upper left')

# Plot training and validation loss
ax[1].plot(history.history['loss'])
ax[1].plot(history.history['val_loss'])
ax[1].set_title('Model Loss')
ax[1].set_xlabel('Epoch')
ax[1].set_ylabel('Loss')
ax[1].legend(['Train', 'Validation'], loc='upper right')

# Save the figure
plt.savefig('plot CNN-2.png', dpi=300, bbox_inches='tight')

plt.tight_layout()
plt.show()
```

**6.4 Receiver Operating Characteristic (ROC) Curve**

```python
# Compute predicted probabilities for each class
y_probs = model.predict(X_test)

# Ensure that the target labels Y_test are in a 2-dimensional format
if len(y_test.shape) == 1:
    Y_test = np.eye(len(np.unique(y_test)))[y_test.astype(int)]

# Compute the ROC curve and AUC score for each class
fpr = dict()
tpr = dict()
roc_auc = dict()
for i in range(Y_test.shape[1]):
    fpr[i], tpr[i], _ = roc_curve(Y_test[:, i], y_probs[:, i])
    roc_auc[i] = roc_auc_score(Y_test[:, i], y_probs[:, i])

# Plot the ROC curve for each class
plt.figure()
for i in range(Y_test.shape[1]):
    plt.plot(fpr[i], tpr[i], label=f'Class {i} (AUC = {roc_auc[i]:.2f})')

# Set the title and axis labels
plt.title('Receiver Operating Characteristic (ROC) Curve')
plt.xlabel('False Positive Rate')
plt.ylabel('True Positive Rate')
plt.legend(loc='lower right')

# Save the figure
plt.savefig('ROC CNN-2.png', dpi=300, bbox_inches='tight')

# Show the plot
plt.show()
```
